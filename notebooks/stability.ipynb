# Homogeneous stability
The Jacobian and Hopf search live in `urbancrime.analysis.stability`; this notebook only loads the Figure 3 configuration and plots one configured sweep.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import yaml
from urbancrime.analysis.stability import StabilityParameters, parameter_sweep

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
with (ROOT / 'configs/smoke/stability.yaml').open() as stream:
    study = yaml.safe_load(stream)
parameters = StabilityParameters(**study['base_parameters'])
search = {
    'max_mode': study['spectral']['max_mode'],
    **study['root_search'],
}
sweep = study['sweeps'][0]
values = np.linspace(sweep['values']['start'], sweep['values']['stop'], sweep['values']['count'])
delays = parameter_sweep(parameters, sweep['parameter'], values, **search)
```

```python
plt.semilogy(values, delays)
plt.xlabel(r'$\eta$')
plt.ylabel(r'$\tau_c^*$')
plt.grid(alpha=0.25);
```
