# PDE quickstart
The notebook contains no model equations or parameters; it loads a checked-in configuration and calls the package runtime.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from urbancrime.config import load_run_config
from urbancrime.runtime import run_pde

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
CONFIG = ROOT / 'configs/smoke/no_police.yaml'
result = run_pde(load_run_config(CONFIG), output_directory=ROOT / 'runs/notebook-quickstart')
```

```python
plt.plot(result.times, result.metrics['crime_average'])
plt.xlabel('time')
plt.ylabel('average crime intensity')
plt.grid(alpha=0.25);
```
